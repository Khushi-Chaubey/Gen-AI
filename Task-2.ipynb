### Task 2: Categories(Sets)

```python
#parallel list of categories matching product_list length
categories = ["Cooking", "Baby Care", "Grocery", "Hygiene", "Grocery", "Personal Care","Personal Care", "Personal Care"]

#Create set from the list
categories_set = set(categories)
print("Initial categories set: ", categories_set)
```

```python
# Adding a new unique category
categories_set.add("Beverages")
print("After adding 'Beverages': ", categories_set)

# Attempting to add an existing category (duplicate)
categories_set.add("Grocery")
print("After trying to add duplicate 'Grocery': ", categories_set)
```

```python
# Check if category exists
is_present = "Hygiene" in categories_set
print("Is 'Hygiene' in categories_set:", is_present)

# Check for a category that does not exist:
is_missing = "Comb" in categories_set
print("Is 'Comb' in categories_set:", is_missing) 
```

```python
# Optional: get total number of unique categories
total_unique = len(categories_set)
print("Total number of unique categories:", total_unique)
```
