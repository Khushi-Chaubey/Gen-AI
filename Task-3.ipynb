### Task 3: Product pricing (Dictionaries)

```python
# Create price_dict with at least 6 product entries
price_dict = {
    "Oil": 150,
    "Diaper": 450,
    "Fruits": 80,
    "Handwash": 65,
    "Snacks": 120,
    "Soap": 30,
    "Shampoo": 200,
    "Bodywash": 180
}
```

```python
print(price_dict)
```

```python
# Add a new product with price
price_dict["Toothpaste"] = 75
print("After adding 'Toothpaste':", price_dict)
```

```python
# Update the price of an existing product
price_dict["Soap"] = 45
print("After updating 'Soap' price to 45:", price_dict)
```

```python
# Remove a product by name with error handling
product_to_remove = "Fruits"
if product_to_remove in price_dict:
    price_dict.pop(product_to_remove)
    print(f"Product '{product_to_remove}' removed successfully")
else:
    print(f"Product '{product_to_remove}' does not exist in the dictionary")

print("After removing 'Fruits':", price_dict)

product_to_remove = "Milk"
if product_to_remove in price_dict:
    price_dict.pop(product_to_remove)
else:
    print(f"Product '{product_to_remove}' does not exist in the dictionary")
```

```python
# Calculate average price using dictionary operations
average_price = sum(price_dict.values()) / len(price_dict)
print(f"Average price of all products: {average_price}")
```

```python
# Find products with maximum and minimum prices
max_product = max(price_dict, key=price_dict.get)
min_product = min(price_dict, key=price_dict.get)
print(f"Product with maximum price: {max_product} at {price_dict[max_product]}")
print(f"Product with minimum price: {min_product} at {price_dict[min_product]}")
```
